```python
decimal = int(input("Enter a decimal number: "))
bits = int(input("Enter the number of bits: "))

binary = bin(decimal)[2:]

if len(binary) < bits:
    binary = "0" * (bits-len(binary))+binary
else:
    binary = binary[-bits:]
    
print(binary)

largest_decimal= 2**bits-1

print(largest_decimal)
```

```python
n = int(input("Enter a positive integer: "))

sum=0
for i in range(1,n+1):
    sum += i
    
print(sum)

formula=n*(n+1)/2

print(formula)

bool(sum==formula)
```

```python
n = int(input("Enter a positive integer: "))

reverse = 0

while n > 0:
    remainder = n % 10
    reverse = reverse*10 + remainder
    n = n // 10
    
print(reverse)
```

```python
even_wo_four = []

for n in range(0,1001):
    if n % 2 ==0 and n % 4 != 0:
        even_wo_four.append(n)
        
print(even_wo_four)


leap_year = []

for i in range(1890,2051):
    if i % 4 == 0 and (i % 100 !=0 or i % 400 == 0):
        leap_year.append(i)
        
print(leap_year)

prime = []

for a in range(2,101):
    prime_number = True
    
    for b in range(2,a):
        if a % b ==0:
            prime_number = False
            break
    
    if prime_number:
        prime.append(a)
        
print(prime)
        
```

```python
n = int(input("Enter a positive integer: "))

while n != 1:   
    if n % 2 == 0:
        n = n // 2

    else:
        n = n * 3 + 1
    
print(n)
```
